```python
df = pd.read_csv("retail_sales_dataset.csv")
print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
display(df.head())
```

```python
category_sales = df.groupby("Product Category")["Total Amount"].sum()
plt.figure(figsize=(8, 5))
plt.bar(
    category_sales.index,
    category_sales.values
)
plt.title("Total Sales by Product Category",
          fontsize=14, fontweight="bold")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
# Convert Date column to datetime
df["Date"] = pd.to_datetime(df["Date"])
# Extract month
df["Month"] = df["Date"].dt.to_period("M")
# Calculate monthly sales
monthly_sales = df.groupby("Month")["Total Amount"].sum()
# Create line chart
plt.figure(figsize=(12, 5))

plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    marker="o",
    linewidth=2
)
plt.title("Monthly Sales Trend", fontsize=15, fontweight="bold")
plt.xlabel("Month")
plt.ylabel("Total Sales Amount")
plt.xticks(rotation=45)
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# Count transactions in each product category
category_count = df["Product Category"].value_counts()
# Create pie chart
plt.figure(figsize=(7, 7))
plt.pie(
    category_count.values,
    labels=category_count.index,
    autopct="%1.1f%%",
    startangle=90,
    shadow=True
)
plt.title("Product Category Distribution", fontsize=15, fontweight="bold")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))
plt.hist(
    df["Total Amount"],
    bins=20,
    edgecolor="black"
)
plt.title("Distribution of Total Sales Amount",
          fontsize=15, fontweight="bold")
plt.xlabel("Total Amount")
plt.ylabel("Number of Transactions")
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))
plt.scatter(
    df["Age"],
    df["Total Amount"],
    alpha=0.7
)
plt.title("Age vs Total Sales Amount",
          fontsize=15, fontweight="bold")
plt.xlabel("Customer Age")
plt.ylabel("Total Amount")
plt.grid(True, linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

```python
# Final Data Visualization Insights
print("===== DATA VISUALIZATION INSIGHTS =====")
# 1. Total sales
total_sales = df["Total Amount"].sum()
print(f"\n1. Total Sales Amount: {total_sales:,.2f}")
# 2. Total transactions
total_transactions = len(df)
print(f"2. Total Transactions: {total_transactions}")
# 3. Average transaction amount
average_sales = df["Total Amount"].mean()
print(f"3. Average Transaction Amount: {average_sales:,.2f}")
# 4. Best-selling category by sales amount
category_sales = df.groupby("Product Category")["Total Amount"].sum()
best_category = category_sales.idxmax()
print(f"4. Highest Revenue Category: {best_category}")
# 5. Category with most transactions
popular_category = df["Product Category"].value_counts().idxmax()
print(f"5. Most Frequent Product Category: {popular_category}")
# 6. Highest sales month
df["Date"] = pd.to_datetime(df["Date"])
monthly_sales = df.groupby(df["Date"].dt.to_period("M"))["Total Amount"].sum()
best_month = monthly_sales.idxmax()
print(f"6. Highest Sales Month: {best_month}")
print("\n===== ANALYSIS COMPLETED SUCCESSFULLY =====")
```
